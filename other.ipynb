```python
import polars as pl
import streamlit as st
from typing import Literal


def get_csv(name: Literal['All', 'Old Testament', 'New Testament', 'Book of Mormon', 'Doctrine and Covenants', 'Pearl of Great Price']):
    if name == 'All':
        return ["data/Book of Mormon.csv"]


def show_works_books():
    work_drop_box = st.multiselect("Select a Standard Work", options=['All', 'Old Testament', 'New Testament', 'Book of Mormon', 'Doctrine and Covenants', 'Pearl of Great Price'])
    book_drop_box = st.multiselect("Select a Book(s) (Optional)")

def search_word(csv_file: str, word: str, book: str | None = None) -> pl.DataFrame:
    df = pl.read_csv(csv_file)
    if book:
        df = df.filter(pl.col('book_title') == book)

    verses = df.filter(pl.col('scripture_text').str.contains(pattern=fr"(?i)\b{word}\b")).select(['verse_title', 'scripture_text'])

    return verses

from semantic.retrieval import Retrieval

retriver = Retrieval('data/bom.csv', 'bom_e.npy')

        
```
