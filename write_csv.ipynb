```python
import polars as pl
```

```python
scriptures = pl.read_csv('data/lds-scriptures.csv')
```

```python
books_list = scriptures.select(pl.col('volume_title').unique()).to_series().to_list()

for book in books_list:
    df = scriptures.filter(pl.col('volume_title') == book)
    df.write_csv(f'data/{book}.csv')
```
